```kotlin
val user = mapOf(
    "name" to listOf("刘洋", "刘文", "刘群"),
    "age" to listOf("18", "20", "18")
)

val names = user["name"] ?: emptyList()
val ages = user["age"] ?: emptyList()

// 方法2：使用 zip 将两个列表配对
val pairs = names.zip(ages)

for ((name, age) in pairs) {
    println("$name : $age")
}
```

```kotlin
val user = mapOf(
    "name" to listOf("刘洋", "刘文", "刘群"),
    "age" to listOf("18", "20", "28")
)


val names = user["name"] ?: emptyList()
val ages = user["age"] ?: emptyList()

val pair = names.zip(ages)
for ((name,age) in pair){
    println("$name : $age")
}
```
