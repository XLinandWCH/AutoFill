```kotlin
// 在 Notebook 的第一个格子里输入
@file:DependsOn("org.jetbrains:markdown:0.7.3")
@file:DependsOn("com.mikepenz:multiplatform-markdown-renderer-m3:0.27.0")

import org.intellij.markdown.flavours.commonmark.CommonMarkFlavourDescriptor
import org.intellij.markdown.parser.MarkdownParser
```

```kotlin
import org.intellij.markdown.html.HtmlGenerator

val text = "# 欢迎学习\n这个子啊NotBook 里面渲染"
val flavour = CommonMarkFlavourDescriptor()
val parsedTree = MarkdownParser(flavour).buildMarkdownTreeFromString(text)
val html = HtmlGenerator(text,parsedTree,flavour).generateHtml()

// noteBook 特殊语法：直接展示 HTML
HTML(html)
```
